## Set up the workspace

For a local Bash kernel, run `task kernel:crop` in a terminal at the repository root, then select **Bash (crop Hatch)** in the notebook kernel picker. The task creates the crop Hatch environment and registers the kernel for your user. Hatch and Python 3.10 or newer must be installed. Adjust `WORKSPACE` to your checkout, then create a folder for the outputs:

```bash
export WORKSPACE=/workspace/mastering-app-package
export RUNTIME="${WORKSPACE}/runs"
export CROP_PROJECT="/data/work/github-com/eoap/mastering-app-package/water-bodies/command-line-tools/crop"
mkdir -p "${RUNTIME}"
```

## Run the crop application 

First dump the help:

```bash
crop --help
```

Now, run the `crop` command line tool with the parameters:

- input-item: https://earth-search.aws.element84.com/v0/collections/sentinel-s2-l2a-cogs/items/S2B_10TFK_20210713_0_L2A
- aoi: `{"type":"Polygon","coordinates":[[[-121.399,39.834],[-120.74,39.834],[-120.74,40.472],[-121.399,40.472],[-121.399,39.834]]]}`
- epsg: EPSG:4326
- band: green 

```bash
crop \
  --input-item "https://earth-search.aws.element84.com/v0/collections/sentinel-s2-l2a-cogs/items/S2B_10TFK_20210713_0_L2A" \
  --aoi '{"type":"Polygon","coordinates":[[[-121.399,39.834],[-120.74,39.834],[-120.74,40.472],[-121.399,40.472],[-121.399,39.834]]]}' \
  --epsg "EPSG:4326" \
  --band green
```

Do the same but for the _nir_ band:

- input-item: https://earth-search.aws.element84.com/v0/collections/sentinel-s2-l2a-cogs/items/S2B_10TFK_20210713_0_L2A
- aoi: `{"type":"Polygon","coordinates":[[[-121.399,39.834],[-120.74,39.834],[-120.74,40.472],[-121.399,40.472],[-121.399,39.834]]]}`
- epsg: EPSG:4326
- band: nir 

```bash
crop \
    --input-item "https://earth-search.aws.element84.com/v0/collections/sentinel-s2-l2a-cogs/items/S2B_10TFK_20210713_0_L2A" \
    --aoi '{"type":"Polygon","coordinates":[[[-121.399,39.834],[-120.74,39.834],[-120.74,40.472],[-121.399,40.472],[-121.399,39.834]]]}' \
    --epsg "EPSG:4326" \
    --band nir
```

List the outputs:

```bash
ls -lh "${RUNTIME}/crop_green.tif" "${RUNTIME}/crop_nir.tif"
```
