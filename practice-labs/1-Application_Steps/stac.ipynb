## Setup the environment

Create a folder for running the steps:

```bash
export WORKSPACE=/workspace/mastering-app-package
export RUNTIME=${WORKSPACE}/runs
mkdir -p ${RUNTIME}
cd ${RUNTIME}
```

## Create the Hatch environment

Hatch installs the package and its dependencies from `pyproject.toml`. Activate the environment before running the tool in the output directory.

```bash
cd "${WORKSPACE}/water-bodies/command-line-tools/stac"
hatch env create default
source "$(hatch env find default)/bin/activate"
cd "${RUNTIME}"
```

## Run the stac application 

First dump the help:

```bash

stac --help
```

```bash
stac \
    --item "https://earth-search.aws.element84.com/v0/collections/sentinel-s2-l2a-cogs/items/S2B_10TFK_20210713_0_L2A" \
    --rasters otsu.tif
```

List the outputs:

```bash
tree ${RUNTIME}
```
