## Setup the environment

Create a folder for running the steps:

```bash
export WORKSPACE="$(git rev-parse --show-toplevel)"
export RUNTIME=${WORKSPACE}/runs
mkdir -p "${RUNTIME}"
cd "${RUNTIME}"
```

## Create a Hatch environment

Python 3.12 or newer and Hatch are required. From the repository root, run `task kernel:install TOOL=stac` and select **Bash (stac Hatch)** in the notebook kernel picker. A regular Bash kernel can also activate the Hatch environment with the following cell. Activation keeps the installed command available after changing to the output directory.


```bash
cd "${WORKSPACE}/water-bodies/command-line-tools/stac"
hatch env create default
source "$(hatch env find default)/bin/activate"
pip install --upgrade "pip @ https://files.pythonhosted.org/packages/3f/06/3e4420302363fbb71be3c7f35ce042f54bfca330e3e66b3a2e3c4cca579f/pip-26.2.2-py3-none-any.whl#sha256=fd7f4dfa8190e014013b9f08f3e973f4e4c26ffd179a204895533a7254813a70" && \
pip install --upgrade "certifi>=2026.10.19" && \
pip install -e .

which python
which stac

cd "${RUNTIME}"
```

```bash

stac --help
```
